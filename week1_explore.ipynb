# Import activity and list message types

The first thing doing below is importing the needed library and reading the message types inside the fit file

```python
from fitparse import FitFile
import pandas as pd

fit_file = FitFile('22105359350_ACTIVITY.fit')

## Defines a set with the message names which are not 'unknown'
msg_types = set (m.name for m in fit_file.messages if m.name and 'unknown' not in m.name)

print (msg_types)

## definition of a get_record  and get_print_records function which also prints the values

def get_print_records (record_type,amount=None):
    all_records = []
    for count,message in enumerate(fit_file.get_messages (record_type)):
        if amount is not None and count >= amount:
            break
        field_content = {
            field.name : field.value
            for field in message if field.name and 'unknown' not in field.name
        }
        all_records.append (field_content)
    for count,record_dict in enumerate(all_records):
    # print (record_dict)
        print (f"Message {count}")
        for key,value in record_dict.items():
            print (f"{key}:{value}")
#    return all_records

def get_records (record_type,amount=None):
    all_records = []
    for count,message in enumerate(fit_file.get_messages (record_type)):
        if amount is not None and count >= amount:
            break
        field_content = {
            field.name : field.value
            for field in message if field.name and 'unknown' not in field.name
        }
        all_records.append (field_content)
    return all_records
```

### Message Types
From the data I found several things. There are many different message types. For us the most importants would most likely be event, record and hrv.

- Event message types include events triggered in the device, like start, stop, pause. And includes if the pause was automatic or not. This are important to recognize gaps in the data.
- The record includes the ride information and is sent every second. 
- HRV is most likely relevant to recognize some kind of stress levels (however I don't know yet how to the values can be used)

The message pattern was almost always record → hrv → gps_metadata → record → hrv → gps_metadata → ...

But around the event event markers there was a different pattern.

event (stop_all) → hrv → hrv → gps_metadata → ... → event (start) → record → ...

Which indicates that hrv and gps_metadata is sent during the pause.

By the end I just put all message types of 'record' in a pandas dataframe where I could see that there are 6708 entries in total with 14 columns, and some not all the entries have all the values.

We spotted 3 data quality issues:
- Timestamp gaps: We could see that not all the timestamps were given with gaps of 41 seconds in some cases, maybe even more. The assumption is that pauses create gaps as no records are sent when the garmin is paused.
- Sensor dropout: We assume that there are some sensor dropouts as we don't heart rate in all of the records (in 8 of 6708 records the heart rate is missing)
- Inconsistent fields: Related with the sensor dropout, we don't have all the values in all the records, producing NaN-filled rows inside the data frame.
